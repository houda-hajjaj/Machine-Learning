# Projet 1 — Système expert pour la résolution des Tours de Hanoï

### Machine Learning — 2026/2027

**Objectif :** concevoir et analyser un système expert à base de règles
utilisant un moteur d'inférence en chaînage avant pour résoudre les
Tours de Hanoï à 3 pics et 3 disques.

## 1. Contexte et objectifs

Les Tours de Hanoï constituent un problème classique de déplacement
sous contraintes. Le problème considéré comporte trois pics et trois
disques de tailles 1, 2 et 3.

Deux contraintes principales doivent être respectées :

1. un seul disque peut être déplacé à chaque étape ;
2. un disque ne peut jamais être placé sur un disque de taille inférieure.

La configuration initiale place les trois disques sur le pic 0 et
l'objectif consiste à les transférer vers le pic 2.

Dans ce projet, la résolution n'est pas obtenue par l'algorithme
récursif classique des Tours de Hanoï. Nous cherchons à construire
un système expert capable de prendre des décisions à partir d'une
base de règles et d'un mécanisme d'inférence en chaînage avant.

L'étude portera donc autant sur la résolution obtenue que sur le
comportement du système expert : sélection des règles, prévention
des boucles, nombre de situations explorées et écart éventuel par
rapport à la solution optimale.
